```python
#import
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn. metrics import accuracy_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
data=pd.read_csv('/content/heart.csv')
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data['target'].value_counts()
```

```python
x=data.drop(columns='target',axis=1)
```

```python
y=data['target']
```

TRAIN TEST SPLIT

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,stratify=y,random_state=3)
```

```python
models=[LogisticRegression(max_iter=1000),SVC(kernel='linear'),KNeighborsClassifier(),RandomForestClassifier()]
```

```python
def model_train_test():
  for model in models:
    model.fit(x_train,y_train)
    predict_x_test=model.predict(x_test)
    accuracy=accuracy_score(y_test,predict_x_test)
    print("accuracy_score =",model,'=',accuracy)
```

```python
model_train_test()
```

crocess validation

```python
def crocess_validation():
  for model in models:
    cv_score=cross_val_score(model,x,y,cv=5)
    cv_score=(sum(cv_score)/len(cv_score))*100
    cv_score=round(cv_score,2)
    print("cv_score",model,"=",cv_score)
```

```python
crocess_validation()
```
