```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
#import data
di=pd.read_csv('/content/diabetes.csv')
```

```python
di.head()
```

```python
di['Outcome'].value_counts()
```

```python
di.groupby('Outcome').mean()
```

```python
scale=StandardScaler()
X=di.drop(columns='Outcome',axis=1)
Y=di['Outcome']
X_standard_data=scale.fit_transform(X)
print(X_standard_data)
```

```python
x_train,x_test,y_train,y_test=train_test_split(X_standard_data,Y, test_size=0.2,stratify=Y,random_state=2)
print(x_train.shape,y_test.shape)
```

```python
#model import svm & SVC
classify=svm.SVC(kernel='linear')
classify.fit(x_train,y_train)
```

```python
#predict and test score
x_predict=classify.predict(x_train)
score=accuracy_score(x_predict,y_train)
print(score)
```

```python
#print output as given in put
input =(5,166,72,19,175,25.8,0.587,51)
final=np.asarray(input).reshape(1,-1)
input1=scale.fit_transform(final)
print(input1)
ans=classify.predict(final)
print(ans)
if(ans[0]==0):
  print('it is not diabatic')
else:
  print('it is diabatic ')
```
